```python
def age_calculation(age):
    if age <= 19 :
        return "Under 20's"
    elif 20 <= age <= 29 :
        return "20's"
    elif 30 <= age <= 39 :
        return "30's"
    elif 40 <= age <= 49 :
        return "40's"
    elif 50 <= age <= 59 :
        return "50's"
    else :
        return "60's"
```

```python
df["age_group"] = [ age_calculation(age) for age in df["age"]]
```

```python
import dash
from dash import html, dcc, Input, Output
import plotly.express as px
import plotly.graph_objects as go
import pandas as pd

# Initialize app
app = dash.Dash(__name__)

# Load the FULL dataset (not the summary)
df = pd.read_csv(r"C:\works\learnings\analysis\customer_shopping_data.csv")

# Data preparation
df['invoice_date'] = pd.to_datetime(df['invoice_date'], format="%d/%m/%Y")
df['total_revenue'] = df['quantity'] * df['price']
df['year_month'] = df['invoice_date'].dt.to_period('M').dt.to_timestamp()
df['day_name'] = df['invoice_date'].dt.day_name()

# Calculate KPIs
total_revenue = df['total_revenue'].sum()
total_customers = df['customer_id'].nunique()
avg_order_value = df.groupby('invoice_no')['total_revenue'].sum().mean()
total_orders = df['invoice_no'].nunique()

# Prepare data for charts
category_revenue = df.groupby('category')['total_revenue'].sum().reset_index()
monthly_revenue = df.groupby('year_month')['total_revenue'].sum().reset_index()
mall_revenue = df.groupby('shopping_mall')['total_revenue'].sum().reset_index()

# App layout
app.layout = html.Div([
    html.Div([
        html.H1("🛍️ Customer Shopping Dashboard", 
                style={'textAlign': 'center', 'color': '#00d4ff', 'marginBottom': '30px'}),
    ]),
    
    # Filters Section
    html.Div([
        html.Div([
            html.Label("Select Mall:", style={'color': 'white', 'fontWeight': 'bold'}),
            dcc.Dropdown(
                id='mall-filter',
                options=[{'label': 'All Malls', 'value': 'All'}] + 
                        [{'label': mall, 'value': mall} for mall in sorted(df['shopping_mall'].unique())],
                value='All',
                style={'width': '300px', 'color': 'black'}
            )
        ], style={'display': 'inline-block', 'marginRight': '20px'}),
        
        html.Div([
            html.Label("Select Category:", style={'color': 'white', 'fontWeight': 'bold'}),
            dcc.Dropdown(
                id='category-filter',
                options=[{'label': 'All Categories', 'value': 'All'}] + 
                        [{'label': cat, 'value': cat} for cat in sorted(df['category'].unique())],
                value='All',
                style={'width': '300px', 'color': 'black'}
            )
        ], style={'display': 'inline-block', 'marginRight': '20px'}),
        
        html.Div([
            html.Label("Date Range:", style={'color': 'white', 'fontWeight': 'bold'}),
            dcc.DatePickerRange(
                id='date-filter',
                start_date=df['invoice_date'].min(),
                end_date=df['invoice_date'].max(),
                display_format='DD/MM/YYYY',
                style={'color': 'black'}
            )
        ], style={'display': 'inline-block'}),
        
    ], style={'padding': '20px', 'backgroundColor': '#1e1e1e', 'marginBottom': '20px'}),
    
    # KPI Cards
    html.Div([
        html.Div([
            html.Div([
                html.H4("💰 Total Revenue", style={'color': '#00d4ff'}),
                html.H2(id='kpi-revenue', children=f"${total_revenue:,.0f}", 
                        style={'color': 'white'})
            ], className='kpi-card', style={
                'backgroundColor': '#2d2d2d', 
                'padding': '20px', 
                'borderRadius': '10px',
                'textAlign': 'center',
                'width': '23%',
                'display': 'inline-block',
                'margin': '1%'
            }),
            
            html.Div([
                html.H4("👥 Total Customers", style={'color': '#00d4ff'}),
                html.H2(id='kpi-customers', children=f"{total_customers:,}", 
                        style={'color': 'white'})
            ], className='kpi-card', style={
                'backgroundColor': '#2d2d2d', 
                'padding': '20px', 
                'borderRadius': '10px',
                'textAlign': 'center',
                'width': '23%',
                'display': 'inline-block',
                'margin': '1%'
            }),
            
            html.Div([
                html.H4("🛒 Total Orders", style={'color': '#00d4ff'}),
                html.H2(id='kpi-orders', children=f"{total_orders:,}", 
                        style={'color': 'white'})
            ], className='kpi-card', style={
                'backgroundColor': '#2d2d2d', 
                'padding': '20px', 
                'borderRadius': '10px',
                'textAlign': 'center',
                'width': '23%',
                'display': 'inline-block',
                'margin': '1%'
            }),
            
            html.Div([
                html.H4("📊 Avg Order Value", style={'color': '#00d4ff'}),
                html.H2(id='kpi-aov', children=f"${avg_order_value:,.2f}", 
                        style={'color': 'white'})
            ], className='kpi-card', style={
                'backgroundColor': '#2d2d2d', 
                'padding': '20px', 
                'borderRadius': '10px',
                'textAlign': 'center',
                'width': '23%',
                'display': 'inline-block',
                'margin': '1%'
            }),
        ])
    ], style={'marginBottom': '30px'}),
    
    # Charts Row 1
    html.Div([
        html.Div([
            dcc.Graph(id='revenue-trend')
        ], style={'width': '48%', 'display': 'inline-block', 'padding': '10px'}),
        
        html.Div([
            dcc.Graph(id='category-chart')
        ], style={'width': '48%', 'display': 'inline-block', 'padding': '10px'}),
    ]),
    
    # Charts Row 2
    html.Div([
        html.Div([
            dcc.Graph(id='mall-chart')
        ], style={'width': '48%', 'display': 'inline-block', 'padding': '10px'}),
        
        html.Div([
            dcc.Graph(id='gender-chart')
        ], style={'width': '48%', 'display': 'inline-block', 'padding': '10px'}),
    ]),
    
], style={'backgroundColor': '#1a1a1a', 'padding': '20px'})

# Callbacks for interactivity
@app.callback(
    [Output('kpi-revenue', 'children'),
     Output('kpi-customers', 'children'),
     Output('kpi-orders', 'children'),
     Output('kpi-aov', 'children'),
     Output('revenue-trend', 'figure'),
     Output('category-chart', 'figure'),
     Output('mall-chart', 'figure'),
     Output('gender-chart', 'figure')],
    [Input('mall-filter', 'value'),
     Input('category-filter', 'value'),
     Input('date-filter', 'start_date'),
     Input('date-filter', 'end_date')]
)
def update_dashboard(selected_mall, selected_category, start_date, end_date):
    # Filter data
    filtered_df = df.copy()
    
    if selected_mall != 'All':
        filtered_df = filtered_df[filtered_df['shopping_mall'] == selected_mall]
    
    if selected_category != 'All':
        filtered_df = filtered_df[filtered_df['category'] == selected_category]
    
    if start_date and end_date:
        filtered_df = filtered_df[
            (filtered_df['invoice_date'] >= start_date) & 
            (filtered_df['invoice_date'] <= end_date)
        ]
    
    # Calculate KPIs
    kpi_revenue = f"${filtered_df['total_revenue'].sum():,.0f}"
    kpi_customers = f"{filtered_df['customer_id'].nunique():,}"
    kpi_orders = f"{filtered_df['invoice_no'].nunique():,}"
    kpi_aov = f"${filtered_df.groupby('invoice_no')['total_revenue'].sum().mean():,.2f}"
    
    # Revenue Trend
    monthly_data = filtered_df.groupby('year_month')['total_revenue'].sum().reset_index()
    fig_trend = px.line(monthly_data, x='year_month', y='total_revenue',
                        title='📈 Revenue Trend Over Time',
                        template='plotly_dark')
    fig_trend.update_traces(line_color='cyan', line_width=3, mode='lines+markers')
    fig_trend.update_xaxes(title_text='Month')
    fig_trend.update_yaxes(title_text='Revenue ($)')
    
    # Category Revenue
    cat_data = filtered_df.groupby('category')['total_revenue'].sum().reset_index()
    cat_data = cat_data.sort_values('total_revenue', ascending=False)
    fig_category = px.bar(cat_data, x='category', y='total_revenue',
                          title='💼 Revenue by Category',
                          template='plotly_dark')
    fig_category.update_traces(marker_color='lightblue')
    
    # Mall Revenue
    mall_data = filtered_df.groupby('shopping_mall')['total_revenue'].sum().reset_index()
    mall_data = mall_data.sort_values('total_revenue', ascending=False).head(10)
    fig_mall = px.bar(mall_data, x='shopping_mall', y='total_revenue',
                      title='🏬 Top Malls by Revenue',
                      template='plotly_dark')
    fig_mall.update_traces(marker_color='orange')
    fig_mall.update_xaxes(tickangle=-45)
    
    # Gender Distribution
    gender_data = filtered_df.groupby('gender')['total_revenue'].sum().reset_index()
    fig_gender = px.pie(gender_data, values='total_revenue', names='gender',
                        title='👫 Revenue by Gender',
                        template='plotly_dark',
                        color_discrete_sequence=[ '#ff6b9d','#00d4ff'])
    
    return kpi_revenue, kpi_customers, kpi_orders, kpi_aov, fig_trend, fig_category, fig_mall, fig_gender

if __name__ == '__main__':
    app.run(debug=True, port=8051)
```
