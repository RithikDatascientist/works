```python
df["total_revnue"] = df["quantity"] * df["price"] * (1 - df["discount"])
df.head(3)
```

```python
avg_order_by_region = df.groupby("region")["total_revnue"].mean()
print(avg_order_by_region)
```

```python
df["avg_order_by_region"] = df["region"].map(avg_order_by_region)
df.head()
```

```python
east_buying_product = df[df["region"] == "East"]["product_category"]
print(east_buying_product)
```

```python
product_mix = df.groupby(["region", "product_category"]).size().reset_index(name = "order_count")
print(product_mix)
```

```python
product_avg = df.groupby("product_category")["total_revnue"].mean().reset_index()
product_avg
```

```python
product_discount = df.groupby(["region","discount", "product_category"]).size().reset_index()
print(product_discount)
```

```python
avg_discount = df.groupby("product_category")["discount"].mean().reset_index()
print(avg_discount)
```

```python
most_sold_product = df.groupby(["region","product_category"])["order_date"].size().reset_index()
print(most_sold_product)
```
