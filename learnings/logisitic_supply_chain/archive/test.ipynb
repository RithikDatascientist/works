```python
df["Month"] = pd.to_datetime(df["Order Date"], dayfirst=True).dt.month
df.head()
```

```python
seasonal_sales_report = df.groupby("Month").agg({
    "Sales" : ["sum", "mean", "median"]
}).reset_index()

seasonal_sales_report.columns = [ "_".join(col) for col in seasonal_sales_report]
seasonal_sales_report
```

```python
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x = seasonal_sales_report["Month_"],
        y = seasonal_sales_report["Sales_sum"],
        mode = "lines+markers"
    )
)

fig.update_layout(
    xaxis_title = "Months in Numbers",
    yaxis_title = "Sales in each month",
    title = "Line chart of monthly sales"
)
fig.show()
```

### Loss Analysis

With the available data (Sales only), it is not possible to determine whether products are sold at a loss, as profitability requires cost or profit information.

### Observations
- Some products show **very low total sales**
- Many have **very low order counts (1–4 orders)**

### Insight
These products are **low-performing in terms of revenue and demand**, but cannot be classified as loss-making.

### Recommendation
To identify loss-making products, additional data is required:
- Profit column, or
- Cost (COGS)

Additionally, for consistency:
- Filter products with higher order counts (e.g., >20) before drawing conclusions
