```python
import numpy as np

X = np.array([
    [1, 2],
    [2, 3],
    [3, 4]
])  # (3 × 2)

w = np.array([0.5, 1.0])  # (2,)

y_pred = X @ w
print(y_pred)
```

```python
p_disease = 0.01
p_pos_given_disease = 0.99
p_pos_given_no_disease = 0.05

p_positive = (
    p_pos_given_disease * p_disease
    + p_pos_given_no_disease * (1 - p_disease)
)

posterior = (p_pos_given_disease * p_disease) / p_positive
posterior
```
