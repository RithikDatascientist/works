# Gradient Descent implementation using only math

```python
# Given : y_hat  = wx+b
#Loss : L = (y-y_hat)^2

x = 6
y = 12
w = 2
b = 1
lr = 0.0001

length = 10
#implement simple Linear regression model 

def graident_descent(x,y,w,b,lr):
    predicted = w*x+b
    new_w = (-2 * (y - predicted)) * x 
    new_b = -2 * (y - predicted)
    w = w-(lr * new_w)
    b = b-(lr * new_b)
    return w,b
for epoch in range(length):
    w,b = graident_descent(x,y,w,b,lr)
    print(w,b)
```

```python
def find_sum_index(arr, target):
    result = []
    for indx, val in enumerate(arr):
        diff = target - val 
        if (diff in arr ) and (diff != val):
            result.append(indx)
    return result
```

```python
print(find_sum_index([2, 7, 11, 15],9))
```

```python
def find_duplicates(arr):
    dec = set()
    duplicate = []
    for i in arr:
        if i not in dec:
            dec.add(i)
        else:
            duplicate.append(i)
    return duplicate
```

```python
print(find_duplicates([1, 2, 3, 2, 4, 5, 3]))
```

```python
def find_moving_avg(arr, k):
    result= []
    l = 0
    while (l+k) <= len(arr):
        val = sum(arr[l:l+k])
        result.append(val/k)
        l+=1
    return result
```

```python
print(find_moving_avg([1, 3, 2, 6, -1, 4, 1, 8, 2],5))
```

```python

def find_anagram(arr):
    dec = {}
    for word in arr:
        p_word = tuple(sorted(word))
        if p_word not in dec :
            dec[p_word] = [word]
        else :
            dec[p_word].append(word)
    return list(dec.values())
            
```

```python
print(find_anagram(["eat","tea","tan","ate","nat","bat"]))
```

```python
def give_score(data):
    avg = df["Regency"].mean()
    if data < avg :
        val =  np.random.randint(4,5, 1)
        return val[0]
    else :
        val = np.random.randint(1,3,1)
        return val[0]
```

```python
df["r_Score"] = pd.qcut(
    df["Regency"],
    q = 5,
    labels=["5", "4", "3", "2", "1"]
)
df.head()
```

```python
df["f_Score"] = pd.qcut(
    df["Frequency"],
    q = 5 ,
    labels=["1", "2", "3", "4", "5"]
)
df.head()
```

```python
df["m_Score"] = pd.qcut(
    df["Monetary"],
    q = 5 ,
    labels=["1", "2", "3", "4", "5"]
)
df.head()
```

```python
def give_segemant(customer):
    row = df.loc[df["customer_id"] == customer].iloc[0]

    r = row["r_Score"]
    f = row["f_Score"]
    m = row["m_Score"]

    if r in ["5", "4"] and f in ["5", "4"] and m in ["5", "4"]:
        return "Champion"
    elif f in ["5", "4"] and  r in ["3", "2", "1"] and m in ["1","2","3"] :
        return "Loyal"
    elif f in ["5", "4"] and m in ["5", "4"] and r in ["1","2","3"] :
        return "High Risk"
    else:
        return "Lost"

print(give_segemant(1))
```

```python
df["Segement"] = df["customer_id"].apply(give_segemant)
df.head()
```

```python
data = [
  {"user_id": 1, "group": "control", "converted": True},
  {"user_id": 2, "group": "treatment", "converted": True},
  {"user_id": 3, "group": "control", "converted": False},
  {"user_id": 4, "group": "treatment", "converted": True}
]
```

```python
def find_conversin_rate(data):
    dec = {}
    result = {}
    for user in data:
        if user["group"] not in dec:
            dec[ user["group"] ] = {}
            dec[ user["group"] ]["user_cnt"] = 1
            dec[ user["group"] ]["converted"] = int(user["converted"])
        else :
            dec[ user["group"] ]["converted"] += int(user["converted"])
            dec[ user["group"] ]["user_cnt"] += 1

    for key,value in dec.items():
        result[key] = {}
        result[key]["conversion_rate"] =  value["converted"]/value["user_cnt"]
    return result
```

```python
print(find_conversin_rate(data))
```
