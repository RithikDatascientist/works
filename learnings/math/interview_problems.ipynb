```python
from nltk.corpus import stopwords
arr = ["  Hello, World!  ", "This is a TEST!!!"]
df = pd.DataFrame(arr, columns=["Words"])
df
```

```python
import string
from nltk.tokenize import word_tokenize
combined_filter = set(stopwords.words("english") + list(string.punctuation))
```

```python
def remove_stop_words(data):
    tokenize = word_tokenize(data)
    filtered = [ w.lower() for w in tokenize if w.lower() not in combined_filter ]
    return " ".join(filtered)
```

```python
df["stop_words_removed"] = df["Words"].apply(remove_stop_words)
```

```python
import re
def feature_extraction(link):
    till_endpoint = link.split("/")
    after_endpoint = re.split(r"[^\w]+",till_endpoint[-1])
    dec = {}
    dec["domain"] = till_endpoint[2]
    dec["path"]  = "/"+after_endpoint[0]
    dec["Params"] = {}
    dec["Params"][after_endpoint[1]] = after_endpoint[2]
    dec["Params"][after_endpoint[3]] = after_endpoint[-1]
    return dec
```

```python
v1 = [10,20,30]
v2 = [10,20,30]

import numpy as np 

def find_similiarity(v1,v2):
    a = np.sqrt(sum( i*i  for i in v1))
    b = np.sqrt(sum( i*i  for i in v2))
    dot = sum(i*j  for i,j in zip(v1,v2))
    return dot/(a*b)
```

```python
print(find_similiarity(v1,v2))
```

```python
def find_mex(arr):
    sort = sorted(arr)
    if len(sort) == 0:
        return 0
    for i in range(len(sort)):
        if i != sort[i]:
            return i
    return max(sort) +1
```

```python
print(find_mex([]))              # Output: 0
print(find_mex([1,2,3,4,5,10]))  # Output: 0
print(find_mex([0,1,2,3,4,5,6])) # Output: 7
print(find_mex([0,1,3,4,1000]))  # Output: 2
print(find_mex([0,2,3,4,5,6]))
```

```python
def parese_passanger_data(data):
    result=[]
    ldata= data.splitlines()
    for row in ldata:
        dec ={}
        temp = row.split(",")
        if len(temp) > 1:
            tec = temp[1].split(".")
            if temp[0] in ["Red","Yellow","Pink"]:
                 dec["color"] = temp[0]
            dec["title"] = tec[0]
            dec["name"] = tec[1].strip()
        else :
            temp = row.split(".")
            dec["title"] = temp[0]
            dec["name"] = temp[1].strip()
        result.append(dec)
                    
    return result

result = parese_passanger_data("""Red, Mr. Owen Harris
Pink, Mrs. John Bradley (Florence Briggs Thayer)
Yellow, Miss. Laina
Red, Mrs. Jacques Heath (Lily May Peel)
Red, Mr. William Henry
Yellow, Mr. James
Mr. Timothy J
Yellow, Master. Gosta Leonard""")
```

```python
import numpy as np

# Dataset
X = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

# Initialize parameters
m = 0.0
b = 0.0
learning_rate = 0.01
epochs = 1000
n = len(X)

# Gradient Descent
for epoch in range(epochs):
    # Predictions
    y_pred = m * X + b
    
    # Calculate loss (MSE)
    loss = (1/n) * np.sum((y - y_pred)**2)
    
    # Calculate gradients
    dm = -(2/n) * np.sum(X * (y - y_pred))
    db = -(2/n) * np.sum(y - y_pred)
    
    # Update parameters
    m = m - learning_rate * dm
    b = b - learning_rate * db
    
    # Print every 100 epochs
    if epoch % 100 == 0:
        print(f"Epoch {epoch}: Loss={loss:.4f}, m={m:.4f}, b={b:.4f}")

print(f"\nFinal: m={m:.4f}, b={b:.4f}")
```
