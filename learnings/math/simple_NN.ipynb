```python
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import accuracy_score,classification_report
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import train_test_split
```

```python
categories = ['alt.atheism','soc.religion.christian','comp.graphics','sci.med']
```

```python
train_data = fetch_20newsgroups(subset="train", categories= categories, remove=("headers","footers","quotes"))
```

```python
test_data = fetch_20newsgroups(subset= "test", categories=categories, remove=("headers","footers","quotes"))
```

```python
print(f"Train df : {len(train_data)}")
print(f"Test df : {len(test_data)}")
print(f"categories : {train_data.target_names}")
```

```python
tfidf = TfidfVectorizer(max_features=3000, stop_words='english')
X_train = tfidf.fit_transform(train_data.data)
X_test = tfidf.transform(test_data.data)
```

```python
print(f"\nTF-IDF shape: {X_train.shape}")
```

```python
# FFNN model
ffnn = MLPClassifier(
    hidden_layer_sizes=(100, 50),
    activation='relu',
    max_iter=30,
    verbose=True,
    random_state=42
)
```

```python
# Train
print("\nTraining FFNN...")
ffnn.fit(X_train, train_data.target)

# Predict
y_pred = ffnn.predict(X_test)

# Evaluate
print(f"\nAccuracy: {accuracy_score(test_data.target, y_pred):.4f}")
print("\nClassification Report:")
print(classification_report(test_data.target, y_pred, 
                          target_names=train_data.target_names))
```

```python
new_text = ["God is love", "Computer graphics are amazing","I dont think gods is real just jesus is true","There is no god","I like to take pills on time."]
new_tfidf = tfidf.transform(new_text)
predictions = ffnn.predict(new_tfidf)
print("\nNew predictions:")
for text, pred in zip(new_text, predictions):
    print(f"'{text}' → {train_data.target_names[pred]}")
```
