```python
import  statsmodels.api as sm
import numpy as np 
```

```python
x = np.array([1,2,3,4,5])
y = np.array([2,4,5,4,5])
```

```python
x = sm.add_constant(x)

model = sm.OLS(y,x).fit()

print(model.summary())
```

**Multi-Model regressor eg walk through**

```python
import pandas as pd 

df = pd.DataFrame({
    'Y' : [ 50,60,70,80,90],
    'x1' : [1,2,3,4,5],
    'x2' :  [2, 1, 4, 3, 5]
}
)

```

```python
X = df[['x1', 'x2']]
X = sm.add_constant(X)  # add intercept
y = df['Y']

model = sm.OLS(y, X).fit()
print(model.summary())
```
