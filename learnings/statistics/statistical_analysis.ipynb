```python
import pandas as pd 
import plotly.graph_objects as go 
import numpy as np 
```

```python
pd.set_option("display.max_columns",None)
```

```python
url = r"C:\works\learnings\statistics\bank-additional\bank-additional\bank-additional-full.csv"
df = pd.read_csv(url, sep=";", header= 0)
df.head()
```

```python
print(df.info())
```

```python
df.describe()
```

```python
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns
stats_df = df[numeric_cols].describe().T
stats_df['variance'] = df[numeric_cols].var()
stats_df["skweness"] = df[numeric_cols].skew()
stats_df["kurtosis"] = df[numeric_cols].kurt()
stats_df.head()
```

**Correlation Matrix**

```python
corr_matrix = df[numeric_cols].corr()

fig = go.Figure(
    data=[
        go.Heatmap(
            z = corr_matrix.values,
            x = corr_matrix.columns,
            y = corr_matrix.columns,
            colorscale="RdBu",
            zmid=0,
            text=corr_matrix.values,
            texttemplate='%{text:.2f}',
            textfont={"size": 10},
            colorbar=dict(title="Correlation")
        )
    ]
)

fig.update_layout(
    title='Correlation Heatmap',
    width=800,
    height=800,
    xaxis_title='Features',
    yaxis_title='Features',
    yaxis={'autorange': 'reversed'}
)

fig.show()
```

```python
import plotly.figure_factory as ff 

for col in numeric_cols :
    fig = ff.create_distplot([df[col].dropna()], [col], show_hist= True , show_rug=False)
    fig.show()
```
