```python
from scipy.stats import chi2_contingency
import numpy as np 


data = np.array([[30,10],[20,40]])

chi2,p,dof,expected = chi2_contingency(data)

print("Chi-square:", chi2)
print("p-value:", p)
print("Degrees of freedom:", dof)
print("Expected frequencies:\n", expected)
```

```python
from scipy.stats import mannwhitneyu 

store_A = [82, 79, 91, 88, 75]
store_B = [69, 73, 68, 74, 70]

u_stat , p = mannwhitneyu(store_A,store_B,alternative='two-sided')
print(f'U Statistics : {u_stat}')
print(f'P_Value : {p:.2f}')
```

```python
from scipy.stats import wilcoxon

before = [7, 6, 8, 9, 5, 7]
after  = [5, 4, 6, 7, 5, 6]

stat, p = wilcoxon(before, after)

print("Wilcoxon statistic:", stat)
print("p-value:", p)
```

```python
from scipy.stats import kruskal

partner_A = [78, 82, 85, 80]
partner_B = [72, 70, 75, 68]
partner_C = [90, 88, 85, 92]

h_stat, p = kruskal(partner_A, partner_B, partner_C)

print("H statistic:", h_stat)
print(f'P_Value : {p:.2f}')
```
