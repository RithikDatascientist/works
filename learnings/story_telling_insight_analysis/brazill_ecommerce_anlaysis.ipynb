```python
order_dataset["order_delivered_customer_date"] = pd.to_datetime(order_dataset["order_delivered_customer_date"])
order_dataset["order_estimated_delivery_date"] = pd.to_datetime(order_dataset["order_estimated_delivery_date"])
```

```python
order_dataset["Delay"] = order_dataset["order_delivered_customer_date"] - order_dataset["order_estimated_delivery_date"]
order_dataset.head()
```

```python
order_dataset["Delay"] = order_dataset["Delay"].dt.total_seconds() / 86400
delayed_order_dataset = order_dataset[order_dataset["Delay"] > 1]
delayed_order_dataset.head()
```

```python
delayed_order_dataset.info()
```

```python
order_item.info()
```

```python
order = delayed_order_dataset.merge(order_item,on="order_id", how="left")
order["Delay"] = np.ceil(order["Delay"]) 
order.head()
```

```python
order.info()
```

```python
len(order[order["Delay"] > 3]["seller_id"])
```

```python
seller_delay = order.groupby("seller_id")["Delay"].agg(["count","mean","max"])
seller_delay = seller_delay.sort_values("count", ascending=False)
seller_delay.head(20)
```
