# Olist Marketplace — Supply Chain Analytics Findings

## Finding 1: Delay is a Seller Concentration Problem
A disproportionate share of delivery failures originates from a critically small seller segment. Just 20 sellers — representing 1.57% of the delayed seller base — are responsible for 24.6% of all delayed orders on the platform. These sellers are not peripheral contributors; they collectively generate 19.3% of total marketplace revenue, creating a direct financial dependency that prevents simple removal. This is not a platform-wide operational failure — it is a concentrated seller performance problem that the business has structurally allowed to persist because of its revenue exposure to these same sellers. The recommended intervention is a 90-day performance remediation program targeting these 20 sellers specifically, with defined delay thresholds and contract exit criteria as consequences for non-compliance.

## Finding 2: Freight Cost Structure is Silently Draining Margins
The marketplace has a systemic freight-to-price ratio problem that is not visible in top-line revenue but is actively eroding order-level profitability. 36.6% of all orders carry freight costs exceeding 30% of the product price — a margin-stressed threshold that makes a large portion of the catalog economically fragile. More critically, 3.72% of orders are being fulfilled at a net loss, where freight costs exceed the product price entirely. The worst freight-ratio sellers show average freight costs 2.5 to 3.5 times the product value. This is a pricing structure failure, not a logistics efficiency problem. The fix requires implementing minimum order value policies and freight-to-price ratio caps by product category to eliminate loss-making order fulfillment.

## Finding 3: Delay and Freight Failures are Structurally Independent — and Compounded by Customer Damage
Cross-referencing the top 20 delay sellers against the top 50 highest freight-ratio sellers reveals zero overlap — these are two entirely separate seller populations driving two entirely separate failure modes simultaneously. The business cannot solve both problems with a single intervention. However, the delay problem carries a customer experience dimension that amplifies its damage beyond operations: 75% of the top 20 delay sellers (15 out of 20) also rank among the worst-reviewed sellers on the platform, and these 20 sellers alone account for 22% of all bad reviews (score < 3) across the entire marketplace. Operational failure is visibly translating into customer dissatisfaction at scale, meaning the revenue dependency on these sellers comes with a compounding hidden cost in customer trust erosion.

## Finding 4: Geographic Delay Patterns Reveal a Logistics Infrastructure Gap
State-level and route-level analysis separates seller-driven delays from infrastructure-driven delays. Rio de Janeiro receives delayed shipments from four different origin states — SC, PR, SP, and MG — with average delays of 14 to 16 days regardless of which seller fulfilled the order. This rules out seller performance as the cause and points directly to last-mile carrier failures into RJ. Similarly, SP→Northeast corridors (CE, SE, PI, RN) consistently average 14 to 15 days of delay across high order volumes, indicating that northeastern Brazil is structurally underserved by the current carrier network. The single worst route, MG→BA, averages 21.6 days. These are not seller problems — they require logistics partner renegotiation, regional carrier expansion, or fulfillment center investment in underserved corridors.
